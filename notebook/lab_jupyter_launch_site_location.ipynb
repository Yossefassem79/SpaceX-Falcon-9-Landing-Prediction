Next, let's create markers for all launch records.

```python
from geopy.distance import geodesic

# Use the existing site_map and launch-site coordinates
# Example launch site: CCAFS SLC-40
launch_lat = 28.563197
launch_lon = -80.576820
launch_location = (launch_lat, launch_lon)

# Example nearby points (replace with actual coordinates from your lab)
nearby_points = {
    "Nearby point 1": (28.565, -80.570),
    "Nearby point 2": (28.560, -80.585),
}

for name, coordinates in nearby_points.items():
    distance_km = geodesic(launch_location, coordinates).km

    folium.Marker(
        location=coordinates,
        popup=f"{name}: {distance_km:.2f} km",
        tooltip=f"{name}: {distance_km:.2f} km",
        icon=folium.Icon(color="blue", icon="info-sign")
    ).add_to(site_map)

    folium.PolyLine(
        locations=[launch_location, coordinates],
        tooltip=f"{distance_km:.2f} km"
    ).add_to(site_map)

site_map
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

*TODO:* Mark down a point on the closest coastline using MousePosition and calculate the distance between the coastline point and the launch site.


```python
# find coordinate of the closet coastline
# e.g.,: Lat: 28.56367  Lon: -80.57163
# distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)

from math import radians, sin, cos, sqrt, atan2

def calculate_distance(lat1, lon1, lat2, lon2):
    R = 6371  # Earth's radius in kilometers

    dlat = radians(lat2 - lat1)
    dlon = radians(lon2 - lon1)

    a = (
        sin(dlat / 2)**2
        + cos(radians(lat1))
        * cos(radians(lat2))
        * sin(dlon / 2)**2
    )

    return 2 * R * atan2(sqrt(a), sqrt(1 - a))


# Launch site: CCAFS SLC-40
launch_site_lat = 28.563197
launch_site_lon = -80.576820

# Selected coastline point (example from the lab)
coastline_lat = 28.56367
coastline_lon = -80.57163

distance_coastline = calculate_distance(
    launch_site_lat,
    launch_site_lon,
    coastline_lat,
    coastline_lon
)

print(f"Distance to coastline: {distance_coastline:.2f} km")


```

```python
# Create and add a folium.Marker on your selected closest coastline point on the map
# Display the distance between coastline point and launch site using the icon property 
# for example
# distance_marker = folium.Marker(
#    coordinate,
#    icon=DivIcon(
#        icon_size=(20,20),
#        icon_anchor=(0,0),
#        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance),
#        )
#    )


from folium.features import DivIcon

coastline_point = [coastline_lat, coastline_lon]

distance_marker = folium.Marker(
    location=coastline_point,
    icon=DivIcon(
        icon_size=(120, 20),
        icon_anchor=(0, 0),
        html=(
            '<div style="font-size: 12px; color: #d35400;">'
            f'<b>{distance_coastline:.2f} KM</b>'
            '</div>'
        )
    ),
    popup="Selected coastline point"
)

distance_marker.add_to(site_map)

# Draw a line between the launch site and coastline point
folium.PolyLine(
    locations=[
        [launch_site_lat, launch_site_lon],
        coastline_point
    ],
    color="red",
    weight=2,
    tooltip=f"Distance: {distance_coastline:.2f} km"
).add_to(site_map)

site_map

```
